## Qué no podemos calcular todavía
La fecha del metadato no siempre es defensa, el marco de selección está incompleto y no hay anotación de uso declarado. La siguiente celda comprueba que no inventamos etiquetas.

```python
from synesis.core import read_jsonl
rows = read_jsonl(ROOT / "reports/features.jsonl")
assert all(r["ai_use_status"] == "not_assessed" for r in rows)
assert not any(r["abstract_text_analyzed"] for r in rows)
print(len(rows), "registros; uso instrumental no evaluado")
```
